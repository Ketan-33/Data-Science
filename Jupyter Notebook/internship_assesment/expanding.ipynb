```python
import pandas as pd
import numpy as np
import json
from datetime import datetime
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('trade_data.csv')

# Function to parse Trade_History
def parse_trade_history(trade_history_str):
    if pd.isna(trade_history_str):  # Handle NaN values
        return None
    try:
        if isinstance(trade_history_str, str):
            return eval(trade_history_str)
        return trade_history_str
    except:
        return None

# Function to expand trade data
def expand_trade_data(df):
    expanded_trades = []
    
    for _, row in df.iterrows():
        if pd.isna(row['Trade_History']):  # Skip NaN values
            continue
        trades = parse_trade_history(row['Trade_History'])
        if trades:
            for trade in trades:
                trade_data = {
                    'Port_ID': row['Port_IDs'],
                    **trade
                }
                expanded_trades.append(trade_data)
    
    return pd.DataFrame(expanded_trades)

# Process and expand the data
expanded_df = expand_trade_data(df)

# Display information about expanded dataset
print("Expanded Dataset Shape:", expanded_df.shape)
print("\nExpanded Dataset Columns:", expanded_df.columns.tolist())
print("\nSample of Expanded Data:")
print(expanded_df.head())

# Convert time to datetime
expanded_df['time'] = pd.to_datetime(expanded_df['time'], unit='ms')

# Basic data validation
print("\nData Types:")
print(expanded_df.dtypes)
print("\nMissing Values:")
print(expanded_df.isnull().sum())

# Save expanded dataset for further analysis
expanded_df.to_csv('expanded_trade_data.csv', index=False)
```
